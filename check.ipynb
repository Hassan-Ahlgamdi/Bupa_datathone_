```python
import pandas as pd
df = pd.read_parquet("data/final_df.parquet")

central = df.groupby("PROVIDER_REGION")["CLAIMS_AMOUNT"].sum()[0]
```

```python
x = df.groupby("PROVIDER_TOWN")["CLAIMS_AMOUNT"].sum().sort_values(ascending=False).reset_index()

RH = x[x["PROVIDER_TOWN"] == "Riyadh"].values[0][1]
```

```python
central / RH
```
